# **Project Summary -**

Terrorism continues to be one of the most complex and devastating global challenges, affecting human lives, economies, and political stability worldwide. This project, titled “United Nations Global Terrorism Analysis (UNGTA)”, explores the patterns, frequency, and severity of terrorist activities across the world using the Global Terrorism Database (GTD). The primary goal of this analysis is to identify significant trends, regional hotspots, and attack patterns that can help policymakers and international agencies design more effective counter-terrorism strategies and resource allocation models.

The dataset used for this analysis comprises information on terrorist incidents recorded between 1970 and 2017, covering over 180,000 global events. Each record contains extensive details, including the date, location, attack type, target type, weapon used, and the number of people killed or wounded. For the purpose of this exploratory data analysis (EDA), the dataset was cleaned and refined to include only the most relevant features such as year, country, region, attack type, target type, weapon type, and casualties. Missing values were handled by imputing zeros for unknown casualty counts, and new derived fields such as total casualties (nkill + nwound) and incident date were created to support trend and time-series analyses.

The exploratory analysis began by examining temporal patterns in terrorist activity. A clear upward trend was observed from the 1970s through the early 2010s, with a sharp increase around 2014, marking the highest number of recorded attacks globally. This surge aligns historically with conflicts in the Middle East and the rise of extremist organizations during that period. Following 2014, a gradual decline in incidents was observed, possibly due to increased global counter-terrorism measures and territorial losses of major terrorist groups.

A country-level analysis revealed that terrorism is highly concentrated in a few nations. Countries such as Iraq, Pakistan, Afghanistan, India, and Nigeria consistently accounted for the majority of global incidents. These nations, located primarily in the Middle East, South Asia, and Sub-Saharan Africa, represent regions where political instability, insurgency, and socio-economic factors have fueled sustained terrorist activities. When analyzed by region, the Middle East & North Africa and South Asia showed the highest number of attacks, while Western Europe and North America reported significantly fewer incidents but with relatively higher casualties in isolated events.

In terms of attack types, Bombing/Explosion emerged as the most common method, followed by Armed Assault and Assassination. However, while bombings were most frequent, Hijackings and Facility/Infrastructure Attacks resulted in higher average casualties, making them more lethal per incident. Similarly, when examining target types, Private Citizens & Property, Military, and Police were the most frequently targeted groups, highlighting a mix of civilian and governmental focus in terrorist motives. Statistical testing (Mann–Whitney U test) confirmed that there were significant differences in casualty distributions between certain attack types, reinforcing that some methods are inherently deadlier than others.

The analysis also revealed that most terrorist events cause relatively few casualties, but a small subset of attacks results in exceptionally high fatalities, indicating a highly skewed distribution. This highlights the importance of preventive intelligence — even a small reduction in the probability of high-impact events could save thousands of lives.

Overall, the project demonstrates how systematic exploratory analysis can uncover valuable insights from complex global datasets. It not only visualizes historical patterns but also provides evidence-based guidance for future research and policy formulation. The findings suggest that targeted preventive strategies, intelligence sharing, and focused counter-terrorism resources in the most affected regions could significantly reduce the global burden of terrorism. Additionally, investing in public safety infrastructure, socio-economic development, and peace-building initiatives in high-risk countries can help address the root causes of violent extremism.

In conclusion, the UNGTA project underscores the role of data-driven analysis in understanding global security dynamics. By combining statistical evidence, visualization, and domain interpretation, this study provides a foundation for proactive decision-making in the pursuit of a safer and more peaceful world.

## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.

# 1. Filling missing values for casualties
data['nkill'] = data['nkill'].fillna(0)
data['nwound'] = data['nwound'].fillna(0)

# 2. Creating a new column 'casualties' = killed + wounded
data['casualties'] = data['nkill'] + data['nwound']

# 3. Creating a 'date' column from year, month, day
def create_date(row):
    try:
        year = int(row['iyear'])
        month = int(row['imonth']) if row['imonth'] > 0 else 1
        day = int(row['iday']) if row['iday'] > 0 else 1
        return pd.Timestamp(year=year, month=month, day=day)
    except:
        return pd.NaT

data['date'] = data.apply(create_date, axis=1)

# 4. Handling missing textual columns
data['city'] = data['city'].fillna("Unknown")
data['country_txt'] = data['country_txt'].fillna("Unknown")
data['region_txt'] = data['region_txt'].fillna("Unknown")

# 5. Dropping irrelevant or highly null columns
threshold = 0.5  # if more than 50% values are missing, drop
data = data.loc[:, data.isnull().mean() < threshold]

print("Final Dataset Shape after Cleaning:", data.shape)
````
